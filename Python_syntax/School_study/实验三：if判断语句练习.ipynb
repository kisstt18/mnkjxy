# 3.4 知识点 4：if‑elif‑else 多分支判断

语法格式：

```python
if 条件1:
    语句1
elif 条件2:
    语句2
elif 条件3:
    语句3
else:
    # 前面所有条件都不成立时执行
    语句N
```

执行规则：从上向下依次判断；**一旦某个条件成立，执行对应代码，后面所有 elif、else 全部跳过，不再判断**。

> 
> 全新业务场景：会员等级判定
> | 消费累计 | 会员等级 |
> |--------|--------|
> |≥5000 | 钻石会员 |
> |3000 ~ 4999 | 黄金会员 |
> |1000 ~ 2999 | 白银会员 |
> |0 ~ 999 | 普通会员 |
> | 小于 0 | 输入金额非法 |

# 【Code】演示示例 4

```python
consume = float(input("请输入累计消费金额："))
if consume < 0:
    print("输入金额非法")
elif consume >= 5000:
    print("钻石会员")
elif consume >= 3000:
    print("黄金会员")
elif consume >= 1000:
    print("白银会员")
else:
    print("普通会员")
```

## 综合题 2【中等】

水果店优惠活动：
消费满 200 元，打 7.5 折；满 100 元，打 9 折；不足 100 元不打折。
输入原始消费金额，先判断金额不能小于 0；金额合法，输出实际应付金额，保留 2 位小数；金额小于 0 输出 “金额输入错误”。

## 综合题 3【偏难，数学几何，结合 math 库】

输入长方形的长和宽。
要求：

1. 如果长或者宽小于等于 0，输出 “边长输入非法，不能构成长方形”；
2. 边长合法，计算长方形对角线长度，使用`math.sqrt()`，输出对角线，保留 2 位小数。

```python
import math
chang=float(input("请输入长方形的长:"))
kuan=float(input("请输入长方形的宽:"))
if chang<=0 or kuan<=0:
    print("边长输入非法，不能构成长方形")
else:
    djx=math.sqrt(chang**2+kuan**2)
    print(f"对角线长度为:{djx:.2f}")
```

# 第三次实验小结

1. `if`单分支：条件成立执行；`if‑else`实现二选一逻辑；`if‑elif‑else`处理多分支业务；
2. `and`全部条件成立为真；`or`任意条件成立即为真，用于组合复杂判断条件；
3. if 嵌套：在 if/else 内部继续写判断；**缩进是核心易错点，每一层缩进对齐**；
4. 高频错误总结：忘记末尾冒号、缩进混乱、把赋值`=`当成判断相等`==`；
5. 现实应用：会员等级、门票计费、登录校验、奖学金评选、几何合法性校验。

1. 使用多个互相独立的`if`语句，和使用`if‑elif‑else`有什么运行结果上的区别？举简单例子说明。
答：
使用多个独立的if语句会容易造成重复代码过多不方便阅读，使用if‑elif‑else的操作空间更多,更有逻辑性，如果数值过大过多的时候使用多分支语句会更简洁更好一点。

举例：
if a<0:
   print（"输入错误"）
elif a>0 and a<18:
   print（"可以好好学习了"）
else:
   print（"可以好好工作了"）
#------------------

if a>0:
   print（"可以好好学习了"）

   print（"可以好好工作了"）
   
2. if 嵌套的时候如果内层代码少缩进 4 个空格，程序运行会出现什么问题
答：会出现运行的语句位置错误，原本是if嵌套内部的语句但是因为少缩进了4个空格，导致软件认为它是跟外面的if判断是一个层级的，这样子内层if判断就不会输出结果。
