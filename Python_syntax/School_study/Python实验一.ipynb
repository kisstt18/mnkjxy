# 1.1 知识点讲解：print () 基础输出

`print()` 是 Python 的输出函数，括号里面放要打印的内容：

1. **输出文字（字符串）**：文字必须放在单引号`' '`或者双引号`" "`里面，引号成对，不能一单一双
2. **输出数字**：数字不需要加引号
3. `print`是英文小写，不能写成 Print / PRINT，Python 区分大小写

```python
#示例代码 1：基础 print 输出
# 输出文字（字符串）
print("Hello Python")
print('Python实验')

# 输出数字
print(2026)
print(3.14)
```

# 1.2 知识点讲解：一行 print 输出多个内容

在同一个 print 括号内，多个内容用**逗号 `,`**隔开，输出时自动用空格分隔。
语法：`print(内容1, 内容2, 内容3)`

```python
#示例代码 2：一行输出多个数据
name = "张三"
score = 85
print("姓名：", name, "成绩：", score)
print(10, 20, 30)
```

# 1.3 知识点讲解：转义字符（换行 \n）

常用转义字符：`\n` 代表**换行**，写在字符串引号内部，实现输出换行。

> 
> 注意：`\n` 只能放在引号里面，外面无效

```python
#示例代码 3：换行输出 \n
print("第一行\n第二行\n第三行")
print("姓名：李四\n年龄：20")
```

# 1.4 知识点讲解：字符串拼接 +

`+` 可以把两段字符串拼接在一起，**只能字符串 + 字符串，不能直接字符串 + 数字**

> 
> 易错点：`"数字"+100` 会报错，数字要转字符串 `str(100)`

```python
#示例代码 4：字符串拼接
# 字符串相加拼接
print("我是" + "学生")

# 数字转字符串后拼接
num = 90
print("分数：" + str(num))
```
