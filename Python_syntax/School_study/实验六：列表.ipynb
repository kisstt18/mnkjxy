## 模块 1：列表的创建与索引访问

### 【Markdown 知识点讲解】

列表`list`是有序可变序列，使用**方括号`[]`**，多个元素用逗号隔开。

- 索引从 0 开始，支持负索引，和字符串规则完全一致
- `list()`函数：可以把字符串、range 对象转为列表

### 【Code 演示示例 1】

```python
# 创建列表
fruit = ["apple","banana","orange"]
print(fruit)

# 通过索引获取元素
print(fruit[0])
print(fruit[-1])

# list()转换
s = "hello"
lst = list(s)
print(lst)

r = list(range(1,6))
print(r)
```

## 模块 2：列表的遍历 for /while

### 【Markdown 知识点讲解】

遍历：把列表里面每一个元素逐个取出来处理。

1. `for 变量 in 列表`：直接拿到元素，写法简洁（优先推荐）
2. while：通过索引下标遍历，需要配合`len()`获取列表长度

### 【Code 演示示例 2】

```python
animal = ["cat","dog","bird"]
# for遍历
for a in animal:
    print(a)

# while索引遍历
i = 0
n = len(animal)
while i < n:
    print(animal[i])
    i = i + 1
```

## 模块 3：向列表增加元素 append /extend/insert

### 【Markdown 知识点讲解】

1. `append(元素)`：**在列表末尾添加 1 个元素**
2. `extend(序列)`：把另一个列表全部元素合并进来
3. `insert(下标,元素)`：在指定索引位置插入元素

> 
> ⚠️区分 append 与 extend：append 把整个列表当成**一个元素**添加；extend 拆开元素合并。

### 【Code 演示示例 3】

```python
lst1 = [1,2,3]
lst1.append(4)
print("append后：",lst1)

lst2 = [1,2]
lst3 = [3,4]
lst2.extend(lst3)
print("extend合并：",lst2)

lst4 = [1,3,4]
lst4.insert(1,2)
print("insert插入：",lst4)
```

## 模块 4：查找与统计 index ()、count ()、in 运算符

### 【Markdown 知识点讲解】

- `元素 in 列表`：判断元素是否存在，返回 True/False
- `list.index(元素)`：返回元素第一次出现的索引；**找不到会报错，建议配合 in 先判断**
- `list.count(元素)`：统计该元素在列表中出现多少次

### 【Code 演示示例 4】

```python
nums = [2,5,2,7,9]
print(2 in nums)
print(nums.index(2))
print(nums.count(2))
```

## 模块 5：删除列表元素 del、pop ()、remove ()

### 【Markdown 知识点讲解】

1. `del 列表[索引]`：按下标删除元素
2. `pop(索引)`：删除并返回该位置元素；不写索引默认删除**最后一个元素**
3. `remove(值)`：删除**第一个匹配该值**的元素；值不存在会报错

### 【Code 演示示例 5】

```python
lst = ["a","b","c","d"]
del lst[1]
print("del删除索引1：",lst)

lst2 = [10,20,30]
out = lst2.pop()
print("pop弹出元素：",out,"剩余列表",lst2)

lst3 = [1,2,2,3]
lst3.remove(2)
print("remove删除第一个2：",lst3)
```

## 模块 6：列表反转与排序 reverse ()、sort ()

### 【Markdown 知识点讲解】

- `list.reverse()`：原地反转列表，改变原列表
- `list.sort()`：原地升序排序；`sort(reverse=True)`实现降序排序

> 
> 注意：reverse ()、sort () 直接修改原列表，没有返回新列表。

### 【Code 演示示例 6】

```python
num = [3,1,4,2]
num.reverse()
print("反转：",num)

num2 = [3,1,4,2]
num2.sort()
print("升序：",num2)

num3 = [3,1,4,2]
num3.sort(reverse=True)
print("降序：",num3)
```

## 综合题 1【简单：录入多名成绩】

题目：编写程序，创建空列表，循环输入 3 个学生成绩，使用 append 存入列表，最后打印全部成绩。
解题思路：

1. 创建空列表`score_list = []`
2. for 循环执行 3 次，每次 input 输入，转 float，append 加入列表
3. 循环结束打印列表

```python
# 在此写代码
score_list= []
for i in range(1,4):
    ipt=float(input(f"请输入第{i}个学生的成绩:"))
    score_list.append(ipt)
print(score_list)
```

## 综合题 2【中等：查找指定元素】

题目：水果列表`fruits=["西瓜","苹果","香蕉","苹果","橙子"]`。
输入要查找的水果名称，如果存在：输出它第一次出现的索引；不存在输出 “列表中没有该水果”,如果列表中没有，则需要添加进去。

> 
> 提示：先使用`in`做存在性判断，再调用 index ()，避免报错。

```python
#综合题2答题区
fruits=["西瓜","苹果","香蕉","苹果","橙子"]
ipt=input("请输入要查找的水果名称:")
if ipt in fruits:
    print("存在")
    print(fruits.index(ipt))
else:
    print("列表中没有该水果")
    fruits.append(ipt)
    print(fruits)
```

## 综合题 3【偏难：过滤大于 60 的分数】

题目：已有分数列表`all_score=[45,78,52,91,66,33]`，遍历列表，把大于等于 60 分的元素收集到新列表 pass_score，最后输出及格的全部分数。
解题思路：

1. 定义空列表`pass_score = []`
2. for 循环遍历 all_score
3. if 判断，如果分数 >=60，append 追加到 pass_score
4. 循环结束打印 pass_score

```python
#综合题3答题区

all_score=[45,78,52,91,66,33]
pass_score=[]
for i in all_score:
    if i>=60:
        pass_score.append(i)
pass_score.sort(reverse=False)
print(pass_score)
```

## 【Markdown】实验小结
