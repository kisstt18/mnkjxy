## 模块 1：字符串索引与分片

### 【知识点讲解 Markdown】

字符串就是一串字符，每个字符都有自己的位置编号（索引）：

- 从左往右：第 1 个字符索引是 **0**，第 2 个是 1，依次类推
- 从右往左：最后 1 个字符索引是 **-1**，倒数第 2 个是 - 2

**分片（切片）**：`字符串[起始:结束]`，取出一段子串，注意：包含起始，不包含结束（左闭右开）。

```python
s = "Python"

# 索引：取单个字符
print(s[0])    # 第1个字符 P
print(s[-1])   # 最后1个字符 n

# 分片：取一段
print(s[0:3])  # 从索引0到2，输出 Pyt
print(s[2:])   # 从索引2到最后，输出 thon
print(s[:3])   # 从开头到索引2，输出 Pyt
```

## 模块 3：find () 查找子串位置

### 【知识点讲解 Markdown】

`字符串.find(子串)`：查找子串在字符串中第一次出现的位置。

- 找到：返回子串开头的索引
- 没找到：返回 **-1**

```python
s = "I like learning Python"
print(s.find("like"))    # 输出2（like从索引2开始）
print(s.find("Java"))    # 输出-1（找不到）
```

## 模块 5：split () 分割字符串 & join () 拼接字符串

### 【知识点讲解 Markdown】

- `字符串.split()`：按空格把字符串切成多个部分，返回列表
- `连接符.join(列表)`：把列表里的字符串用连接符拼起来

### 【示例代码】

```python
# split分割
s = "This is a book"
parts = s.split()
print(parts)    # ['This', 'is', 'a', 'book']

# join拼接
new_s = "-".join(parts)
print(new_s)    # This-is-a-book
```

## 模块 7：strip () 去除首尾空格 & lower ()/upper () 大小写转换

### 【知识点讲解 Markdown】

- `字符串.strip()`：去掉字符串开头和结尾的空格（中间空格不管）
- `字符串.lower()`：全部转小写
- `字符串.upper()`：全部转大写

### 【示例代码】

```python
s = "  Hello World  "
print(s.strip())       # Hello World（首尾空格没了）

s2 = "Python"
print(s2.lower())      # python
print(s2.upper())      # PYTHON
```

## 模块 8：format () 格式化输出

### 【知识点讲解 Markdown】

`format()`：把变量填到字符串的大括号 `{}` 里，实现格式化输出。
基本用法： `"文字{}文字".format(变量)`

### 【示例代码】

```python
name = "小明"
age = 20
score = 92.5
print("我叫{}，今年{}岁，考试分数是{:.1f}".format(name, age, score))
# 输出：我叫小明，今年20岁，考试分数是92.5
# :.1f 表示保留1位小数
```

## 综合题 1【简单：用户名格式检查】

题目：设计一个简单的注册检查程序。
要求：

1. 输入用户名，先去掉首尾空格
2. 判断用户名长度是否在 3~10 个字符之间
3. 长度合格输出 "用户名可用"，否则输出 "用户名长度不符"

解题思路：

1. input 获取用户名 → strip () 去空格

3. if 判断长度范围

```python
name=input("请输入您的用户名：")
new_name=name.strip()
if len(new_name)<3 or len(new_name)>10:
    print("用户名长度不符")
else:
    print("用户名可用")
    
```

## 综合题 3【偏难：简易密码强度检测】

题目：写一个密码强度检测小程序。
要求：

1. 输入密码
2. 判断密码长度是否≥8 位，不够 8 位输出 "密码太短，强度弱"
3. 长度够的话，判断密码里是否同时包含字母和数字（提示：用.isalnum () 配合思路，或者用循环 + 条件判断）
4. 简单版实现：如果密码里有大写字母又有数字，输出 "密码强度强"；否则输出 "密码强度中等"

> 
> 提示：可以先把密码转成大写判断有没有字母，再用 isdigit () 判断有没有数字。
> 简化思路：
> 
> 
> - 用 `password.isdigit()` 判断是不是全数字
> - 用 `password.isalpha()` 判断是不是全字母

```python
password=input("请输入您的密码：")
if len(password)>=8:
    has_upper = False
    has_digit = False
    for c in password:
        if c.isupper():
            has_upper = True
        if c.isdigit():
            has_digit = True
    if has_upper and has_digit:
        print("密码强度强")
    else:
        print("密码强度中等")
else:
    print("密码太短，强度弱")
```

1. `s[0:3]` 和 `s[:3]` 结果一样吗？为什么？
2. `strip()` 能去掉字符串中间的空格吗？如果想去掉中间空格用什么方法？

```python
1.答：s[0:3] 和 s[:3]结果一样。因为都是从字符串的下标从0开始取到2不包含3。
2.答：strip（）不能去掉字符串中间的空格。如果想去掉可用使用.split（）把一整个字符串分割成多个小字符串，
然后使用"".join（）把字符串连接起来就可以去除中间的空格。
```
