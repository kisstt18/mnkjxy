## 模块 3：math.pi 圆周率

### ✅ 知识点讲解（Markdown Cell）

`math.pi`：内置高精度圆周率 π ≈3.1415926535，用于圆、球相关计算。
公式：圆面积 \(S=\pi r^2\)；圆周长 \(C=2\pi r\)

```python
#演示代码
import math
r = 5
area = math.pi * r ** 2
print(f"半径{r}的圆面积 = {area:.2f}") # :.2f 保留2位小数输出
```

## 模块 4：math 幂函数、取整函数 math.ceil/math.floor

### ✅ 知识点讲解（Markdown Cell）

## 模块 5：math 三角函数基础（math.sin、math.cos）

⚠️ math 库三角函数**参数是弧度，不是角度**

## 综合题 3【中等：鸡兔同笼（代数求解，**不用循环判断**）】

> 
> 数学背景：鸡兔同笼。已知：总头数 H，总脚数 F。
> 公式推导：
> 设鸡数量 x，兔数量 y
> x + y = H
> 2x +4y = F
> 解得：兔 y = (F - 2*H)/2；鸡 x = H - y
> 场景：养殖场清点，输入总头数、总脚数，计算鸡和兔子分别多少只。
> 提示：输入为整数，使用 int () 转换。

```python
# 学生在这里写代码
H=int(input("请输入总头数"))
F=int(input("请输入总脚数"))
y=(F-2*H)/2
x=H-y
print(f"鸡{int(x)}只，兔{int(y)}只")
```

## 综合题 4【偏难：勾股定理，生活场景】

> 
> 场景：梯子靠墙。墙底到梯子底部水平距离 a，梯子竖直高度 b，求梯子长度（斜边）。
> 勾股定理：(c=sqrt{a^2+b^2})，使用 math.sqrt
> 输入 a,b，输出梯子长度。

```python
# 学生代码区
import math
a=int(input("请输入底长："))
b=int(input("请输入高度:"))
c=math.sqrt(a**2+b**2)
print(f"梯子的长度为:{c}")
```
