## 模块 1：while 循环基础

### 【知识点讲解单元格 Markdown】

while 循环逻辑：**条件成立就重复执行代码块，条件不成立，循环结束**
语法格式：

```python
while 条件表达式:
    循环体代码
```

### 【示例代码单元格】

题目：输出 1~5 的数字

```python
i = 1
while i <= 5:
    print(i)
    i = i + 1   # 计数器自增，非常关键
```

### 【随堂练习 1】

任务：使用 while 循环，输出 5、4、3、2、1（倒序输出）
👉 提示：初始值设为 5，每次减 1

```python
i=5
while(i>=1):
    print(i)
    i-=1

```

## 模块 2：for 循环 + range ()

### 【知识点讲解单元格 Markdown】

for 循环适合：**知道循环要执行多少次**的场景
range 用法：

- range (n)：0 到 n-1
- range (a,b)：a 到 b-1
- range (a,b,step)：a 到 b-1，步长为 step

### 【示例代码单元格】

题目：for 循环输出 1~10 中所有奇数

```python
for i in range(1,11,2):
    print(i)
```

### 【随堂练习 2】

任务：用 for 循环，输出 1~20 里面所有 3 的倍数
👉 提示：range 起始 1，结束 21，步长 3，或者循环内 if 判断

```python
for i in range(1,21):
    if(i%3==0):
        print(i,end=" ")
```

## 模块 3：break 跳出循环

### 【知识点讲解单元格 Markdown】

`break`：一旦执行，**直接终止整个循环**，循环不再继续
经常搭配 if 判断一起使用

### 【示例代码单元格】

不断输入数字，当输入数字 99 时，停止程序

```python
while True:
    num = int(input("请输入数字，输入99结束："))
    if num == 99:
        print("收到结束指令，循环停止")
        break
    print(f"你输入的数字是：{num}")
```

### 【随堂练习 3】

任务：for 循环遍历 1~10，遇到数字 6 就直接结束循环

```python
for i in range(1,11):
    if(i==6):
        break
    print(i,end=" ")
```

## 模块 4：continue 跳过本次循环

### 【知识点讲解单元格 Markdown】

`continue`：跳过**当前这一轮**剩下代码，直接进入下一次循环判断，不会终止整个循环

### 【示例代码单元格】

输出 1~10，跳过数字 4

```python
for i in range(1,11):
    if i == 4:
        continue
    print(i)
```

### 【随堂练习 4】

任务：while 循环输出 1~10，跳过 7，其余数字打印

```python
i=1
while(i<=10):
    if i==7:
        i+=1
        continue
    print(i,end=" ")
    i+=1
```

## 模块 5：循环嵌套（简单双层循环）

### 【知识点讲解单元格 Markdown】

循环里面再写循环，外层循环走一次，内层循环完整跑完一轮

> 
> 注意：缩进是 Python 的灵魂

### 【示例代码单元格】

打印简单矩形：3 行，每行输出 4 个 *

```python
for i in range(3):
    for j in range(4):
        print("*", end="")
    print()
```

### 【随堂练习 5】

任务：打印 4 行，每行依次有 1、2、3、4 个星号

```python
for i in range(1,5):
    for j in range(i):
        print("*",end=" ")
    print()
```

## 综合题 1：累加求和（基础）

题目：让用户输入一个正整数 n，计算 1+2+3+…+n 的总和并输出
解题思路：

1. input 获取 n，转 int
2. 设置总和变量 sum_total = 0
3. 使用循环依次取出 1~n，累加到 sum_total
4. 循环结束后打印结果

```python
sum_total=0
n=int(input("请输入一个正整数："))

for n in range(1,n+1):
    sum_total+=n
print(f"1+2+3+...+{n}的总和为：",sum_total)
```

## 综合题 2：密码验证小游戏（猜密码）

题目：预设密码是 2026，用户可以反复输入密码，答对提示成功并退出；答错提示重新输入，最多允许输入 5 次，5 次全部错误自动结束游戏
解题思路：

1. 设置正确密码、计数器记录输入次数
2. while 循环，次数没到 5 次就可以继续输入
3. if 判断：密码正确 → break；密码错误 → 次数 + 1，提示剩余次数
4. 循环结束判断是猜对了还是次数用完

```python
count=0
while(count<5):
    password = int(input("请输入密码:"))
    if (password ==2026):
        print("密码输入成功")
        break
    else:
        count+=1
        ts=5-count
        if ts>0:
            print(f"密码错误，还有{ts}次机会")
else:
    print("五次全部错误自动结束游戏")
        
```

## 综合题 3：数字猜谜游戏（经典小游戏）

题目：程序自动随机生成一个 1~100 之间的整数作为目标数字，玩家反复输入猜测数字，程序提示：猜大了 / 猜小了 / 猜对了，如果猜的次数等于5次，则判负，并告诉玩家“次数用完，你输了！QAQ”。
猜对结束游戏，统计一共猜了多少次。
解题思路：

1. 导入 random 库，使用`random.randint(1,100)`生成 1 到 100 的随机整数作为答案
2. 计数变量`guess_num = 0`，记录猜测次数
3. while True循环持续接收用户输入
4. 每次输入，计数 + 1，if 比较大小给出提示，内嵌if判断次数是否达到5次跳出循环。
5. 猜对执行 break，最后输出猜测总次数

```python
import random
i=random.randint(1,100)
guess_num=0
while(True):
    sr=int(input("请输入您猜测的数字："))
    guess_num+=1
    if guess_num<5:
        if(sr>i):
            print("猜大了")
        elif(sr<i):
            print("猜小了")
        else:
            print("猜对了")
            print(f"本次猜测的总次数为：{guess_num}次")
            break
            
    else:
        print("次数用完，你输了！QAQ")
        print("正确数字为：",i)
        break


    
```

## 综合题 4：正整数因数查找（数学应用题）

题目：输入一个大于 1 的正整数，找出它所有的因数
解题思路：

1. 获取数字 n
2. for 循环从 1 遍历到 n
3. if 判断 n % i == 0，满足就打印 i（是因数）

```python
n=int(input("请输入一个大于1的正整数："))
for i in range(1,n+1):
    if n%i==0:
        print(f"{i}是因数")
```

## 综合题 5：简易计分统计（贴近生活场景）

题目：连续输入多名学生分数，输入 -1 代表录入结束；统计一共有多少个分数、总分、平均分。分数范围合法 0~100，如果输入超出范围提示无效，不计入统计。
解题思路：

1. 总分、人数初始化为 0
2. while True 循环接收分数
3. break 条件：输入 - 1
4. if 判断分数是否在 0~100 区间，合法才累加、计数
5. 退出循环后计算平均分并输出

```python
sum_score=0
count=0
while(True):
    score=float(input("请输入学生的分数:"))
    if score==-1:
        break
    if score>100 or score<0:
        print("无效分数")
        continue
    count+=1
    sum_score+=score
    print(f"共有{count}个分数")
if count>0:
    avg=sum_score/count
    print("平均分为:",avg)
else:
    print("没有录入有效分数")
```

## 实验思考题（课后）

1. while True 这种无限循环写法，什么时候适合使用？必须依靠什么来停止？
    答：while True适合循环次数预先未知时使用，无法依靠条件自动终止，必须在循环内部用break配合判断语句退出。
2. break 和 continue 的区别是什么？
    答：break直接跳出整个循环；continue跳过本次循环剩余代码，回到循环开头，二者对循环的终止作用完全不同。
3. 双层循环执行顺序是什么？试着修改星号打印代码，输出倒三角。
    答：双层循环先执行外层，在跑完一轮内层；倒三角打印；外层控制行数，内层按行递减输出星号。
